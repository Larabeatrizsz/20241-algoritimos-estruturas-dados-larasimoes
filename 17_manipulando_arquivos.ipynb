```python
#escreveu em um arquivo
arquivo = open('numeros.txt', 'w')
for linha in range(1,101):
  arquivo.write('%d\n' % linha)
arquivo.close()
```

```python
#leitura de um arquivo
arquivo = open('numeros.txt', 'r')
texto = arquivo.read()
print(texto)
arquivo.close()
```
